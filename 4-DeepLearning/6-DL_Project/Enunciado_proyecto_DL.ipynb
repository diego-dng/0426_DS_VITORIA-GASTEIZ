# Enunciado del Proyecto: Resolver Sudoku desde una Foto

## Objetivo
Desarrollar una aplicación que, dada una fotografía de un Sudoku, sea capaz de solucionarlo.

* **Miembros**: El proyecto es individual
* **Datos**: Conjunto de imagenes de Sudokus, y registro de partidas.
* **Fecha de entrega**: Se tendra que entregar un repositorio de proyecto antes de las 9:00 del **Martes 23 de Junio**. Se presentara ese mismo dia con una pequeña demo. No hace falta presentación.

Todo dataset necesario para el modelo sera entregado por los profesores.
