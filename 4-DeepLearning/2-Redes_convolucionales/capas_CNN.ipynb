# Tipos de capas de una Red Convolucional (CNN)

Guía de referencia de las capas que se usan al construir una CNN en Keras.
Para cada una verás **qué hace**, **qué parámetros** recibe y **cuándo usarla**.

> Recuerda la estructura típica de una CNN:
> `Conv + Pooling` (extraen características) → `Flatten` (conecta) → `Dense` (clasifica).


```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

## 1. `Conv2D` — Capa convolucional

**Qué hace:** es el corazón de la CNN. Desliza pequeños filtros por la imagen
buscando patrones (bordes, texturas, formas). Cada filtro produce un "mapa de
características" que indica dónde aparece ese patrón.

```python
layers.Conv2D(32, (3, 3), activation='relu', padding='same')
```

**Parámetros principales:**

| Parámetro | Qué significa | Valores típicos |
|-----------|---------------|-----------------|
| `filters` (1er número) | Cuántos patrones distintos busca la capa | 32, 64, 128, 256 |
| `kernel_size` | Tamaño del filtro que se desliza | `(3, 3)`, `(5, 5)` |
| `activation` | Función de activación tras el filtro | `'relu'` (lo habitual) |
| `padding` | Qué hacer en los bordes de la imagen | `'same'` o `'valid'` |
| `strides` | Cuántos píxeles salta el filtro al desplazarse | `(1, 1)` por defecto |
| `input_shape` | Forma de la entrada (SOLO en la 1ª capa) | `(64, 64, 3)` |

**Sobre `padding`:**
- `'same'`: añade un borde para que el mapa de salida tenga el **mismo tamaño** que la entrada.
- `'valid'`: no añade borde, así que el mapa de salida es **un poco más pequeño**.

**Cuándo usarla:** siempre, en la parte de extracción de características.
Lo normal es ir **aumentando los filtros** según se profundiza (32 → 64 → 128):
las primeras capas detectan cosas simples y las últimas, patrones complejos.

## 2. `MaxPooling2D` — Capa de reducción (pooling)

**Qué hace:** reduce el tamaño de los mapas de características quedándose con el
valor más alto de cada ventana. Conserva lo importante y descarta el detalle fino.
Esto acelera el entrenamiento y reduce el número de parámetros.

```python
layers.MaxPooling2D((2, 2))
```

**Parámetros principales:**

| Parámetro | Qué significa | Valores típicos |
|-----------|---------------|-----------------|
| `pool_size` | Tamaño de la ventana que se resume | `(2, 2)` (lo habitual) |
| `strides` | Cuánto se desplaza la ventana | igual que `pool_size` por defecto |
| `padding` | Cómo tratar los bordes | `'valid'` por defecto |

Con `(2, 2)` el mapa se reduce a la mitad de ancho y alto (de 64×64 pasa a 32×32).

**Cuándo usarla:** normalmente **después de cada `Conv2D`** (o cada pocas).
Existe también `AveragePooling2D`, que coge la media en vez del máximo, pero
`MaxPooling2D` es la opción por defecto y la más usada.

## `Flatten` — Capa de aplanado

**Qué hace:** convierte los mapas de características (que son datos en 2D/3D) en un
**vector de una sola fila**. Es el puente entre la parte convolucional (que trabaja
con imágenes) y la parte densa (que trabaja con vectores).

**Cómo se escribe:**

```python
layers.Flatten()
```

**Parámetros:** no necesita ninguno.

**Cuándo usarla:** una sola vez, **justo antes de las primeras capas `Dense`**.
