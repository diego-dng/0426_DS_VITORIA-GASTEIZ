# Ejercicio Práctico: Diseño y Planificación de tu Primer Agente de IA

## Objetivo del Ejercicio
El error más común al empezar a desarrollar **Agentes de Inteligencia Artificial** es saltar directamente a escribir código sin realizar una planificación previa. A diferencia de una llamada simple a un LLM (donde enviamos un prompt y recibimos una respuesta única), un agente requiere una **arquitectura de autonomía**: debe ser capaz de tomar decisiones iterativas, interactuar con herramientas externas, mantener un estado de memoria y reaccionar ante los cambios de su entorno.

En este ejercicio trabajarás de forma puramente analítica y conceptual. Tu objetivo es **diseñar la arquitectura, lógica y estructura previa** para una idea de Agente de IA antes de codificarlo. Esto te permitirá entender los componentes críticos que diferencian a un script de automatización tradicional de un agente inteligente autónomo.


---

## El Escenario: Selecciona tu Proyecto
Agentes de ejemplo:

* **Ejemplo A — El Agente Inmobiliario:** Un agente autónomo encargado de recibir correos de potenciales inquilinos, analizar si cumplen con los requisitos mínimos de ingresos, buscar propiedades idóneas en una base de datos y proponer un horario de visitas guardándolo en un calendario.
* **Ejemplo B — El Agente  DevOps:** Un agente que monitoriza de forma continua un servidor web. Si detecta un error `500 Internal Server Error`, debe leer de forma autónoma los últimos logs, clasificar la gravedad de la incidencia, buscar soluciones comunes y decidir si aplica un parche automático (ej. reiniciar un servicio) o escala el caso a un humano.
* **Ejemplo C — El Agente "Researcher" de Mercado:** Un agente que recibe un tema de investigación complejo (ej. *"Situación regulatoria del hidrógeno verde en Chile"*), busca fuentes en internet de forma iterativa, extrae datos, descarta desinformación o duplicados, y compila un informe final estructurado de forma 100% independiente.

## Estructura del Ejercicio (Fases a Desarrollar)

Debes crear un documento de diseño técnico que cubra los siguientes 5 bloques conceptuales clave para tu agente seleccionado:

### Fase 1: Identidad y Misión del Agente (El "Core")
Define los fundamentos cognitivos de la IA. No basta con decirle "eres un asistente útil". Debes estructurar con precisión:
1.  **Nombre del Agente:** Un nombre operativo para tu sistema.
2.  **Objetivo Principal (Goal):** Define la condición exacta bajo la cual el agente considerará que ha cumplido con éxito su ciclo de trabajo. ¿Cuándo se detiene?
3.  **Restricciones y Límites (Guardrails):** Especifica qué cosas **jamás** tiene permitido hacer el agente (ej. gastar más de cierta cantidad en llamadas a APIs por ejecución, inventar datos inexistentes, tomar decisiones financieras sin confirmación).
4.  **Borrador de System Prompt Avanzado:** Escribe el prompt del sistema que gobernará al modelo, estructurado limpiamente por bloques claros: *Rol / Contexto*, *Misión principal*, *Reglas obligatorias* y *Formato de salida esperado*.

### Fase 2: Percepción y Entorno (Environment)
Un agente no vive en el aire; interactúa con un ecosistema. Mapea el entorno de tu agente respondiendo a:
1.  **Definición del Entorno:** ¿Cuál es el "mundo" donde opera el agente? (Ej. una base de datos Postgres, una API de Gmail, carpetas del sistema de archivos local, el terminal de comandos).
2.  **Percepción (Inputs):** ¿Cómo se entera el agente de que tiene trabajo que hacer? ¿Cómo lee los cambios del entorno?
3.  **Modelo de Memoria:** Los agentes necesitan recordar para razonar. Explica qué guardará tu agente en su **Memoria a Corto Plazo** (contexto de la sesión actual) y qué persistirá en su **Memoria a Largo Plazo** (historial entre días de trabajo, preferencias del usuario, bases de datos vectoriales).

### Fase 3: Catálogo de Herramientas (Tools / Functions)
Los LLMs por sí mismos solo saben procesar texto; para ser agentes necesitan herramientas. Diseña la estructura conceptual de **tres herramientas específicas** que tu agente podrá invocar. Para cada una debes detallar la estructura exacta que el LLM necesita ver para decidir usarla:
* **Nombre técnico de la función** (ej. `buscar_inmuebles_por_precio`).
* **Descripción en lenguaje natural para el LLM** (Crucial: la explicación exacta de *para qué* y *cuándo* debe llamarla el modelo).
* **Parámetros de entrada:** Nombre de la variable, tipo de dato (string, int, boolean) y descripción de qué contiene.
* **Output o Retorno esperado:** Qué datos le devolverá el sistema al agente tras ejecutarla.

### Fase 4: El Ciclo de Razonamiento (Loop ReAct)
Diseña la secuencia lógica que seguirá el agente utilizando el patrón **ReAct (Reasoning + Acting)**. Describe paso a paso o mediante un flujo de texto cómo interactuará recursivamente la IA:
1.  **Pensamiento (Thought):** ¿Cómo analiza la situación inicial?
2.  **Acción (Act):** ¿Cómo selecciona e invoca la herramienta idónea del catálogo anterior?
3.  **Observación (Observe):** Cuando la herramienta le devuelva un resultado (sea un éxito o un error), ¿cómo asimila el agente esa nueva información antes de dar el siguiente paso?
4.  ¿Cómo detecta que ya tiene la respuesta final y puede romper el bucle?

### Fase 5: Gestión de Riesgos y Control
Los agentes autónomos pueden descontrolarse en entornos de producción. Diseña tu estrategia de contención:
1.  **Control de Bucles Infinitos:** ¿Cómo evitarás que el agente ejecute una misma herramienta cíclicamente sin llegar a ninguna parte si algo falla? (Ej. Límite de pasos máximos `max_iterations`).
2.  **Manejo de Errores Críticos:** Si una herramienta devuelve un error inesperado (ej. la base de datos está caída o la API no responde), ¿qué instrucción o lógica tiene el agente para no colapsar?
3.  **Intervención Humana (Human-in-the-Loop):** Define el punto exacto e innegociable de la ejecución donde el agente **debe congelar su estado** y esperar la aprobación explícita de un supervisor humano antes de continuar.
