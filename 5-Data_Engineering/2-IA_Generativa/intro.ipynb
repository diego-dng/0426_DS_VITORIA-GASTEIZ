# Guía paso a paso: Instalación y uso de OpenRouter

OpenRouter es un enrutador de APIs que nos permite conectar con múltiples modelos de Inteligencia Artificial (LLMs) usando un solo punto de acceso y una única API Key. 

El método más estándar y robusto en Python es utilizar la librería oficial de `openai` y simplemente cambiar la URL base para que apunte a los servidores de OpenRouter.

## Paso 1: Instalación de librerías
Instalar OpenRouter

## Paso 2: Configuración de las credenciales

Para usar OpenRouter, necesitas una API Key. Puedes obtenerla creando una cuenta gratuita en [openrouter.ai](https://openrouter.ai/). 

Una vez la tengas, por seguridad, no debemos escribirla directamente en el código. Usaremos variables de entorno, aunque en Google Colab puedes usar el panel lateral de "Secretos" o simplemente definirla en una variable (solo para pruebas locales).

```python
# set OPENROUTER_API_KEY=----VUESTRA_API_KEY-----
```

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

# Inicializamos el cliente apuntando a OpenRouter en lugar de OpenAI
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)
```

## Paso 3: Hacer nuestra primera petición

Ahora vamos a interactuar con un modelo. La gran ventaja de OpenRouter es que podemos elegir el modelo simplemente cambiando el parámetro `model`. 

Vamos a probar con un modelo gratuito como `qwen3-next-80b` o `meta-llama/llama-3-8b-instruct:free` (que no consumen saldo), pero puedes usar `anthropic/claude-3.5-sonnet` o cualquier otro del catálogo.

```python
modelo = "google/gemma-4-31b-it:free"
#modelo = "nvidia/nemotron-3-ultra-550b-a55b:free"
```

```python
from openrouter import OpenRouter
import os
      
with OpenRouter(
    api_key=os.getenv("OPENROUTER_API_KEY", ""),
    ) as client:
        response = client.chat.send(
        model=modelo,
        messages=[
        {
        "role": "user",
        "content": "donde esta madrid="
        }
        ]
)

print(response.choices[0].message.content)
```
