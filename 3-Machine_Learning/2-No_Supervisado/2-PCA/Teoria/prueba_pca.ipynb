# Imports

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
df = pd.DataFrame(data['data'], columns=data['feature_names'])
df['target'] = data['target']
df
```

# Analisis.

```python
df.info()
```

```python
df["target"].value_counts()
```

```python
columnas = ["mean radius", "mean perimeter", "mean area", "mean texture"]

sns.pairplot(df[columnas])
```

```python
sns.heatmap(df.corr(), vmin=-1, cmap= "coolwarm")
```

# Modelos.

```python
X = df.drop(columns="target")
y = df["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)
```

```python
X_train.shape
```

## PCA

```python
model_pca = PCA(n_components=3) 
X_train_pca = model_pca.fit_transform(X_train)
```

```python
X_train_pca.shape
```

## LogisticRegression

```python
model_lr = LogisticRegression(max_iter=100)
model_lr.fit(X_train_pca, y_train)

model_lr.score(X_train_pca, y_train)
```

## Pipeline de StandarScaler, PCA Y LogisticRegression

```python
pipe = make_pipeline(StandardScaler(), PCA(n_components=3), LogisticRegression(max_iter=1000))
pipe.fit(X_train, y_train)
```

# Evaluaciones.

## LogisticRegression.


## PCA y LogisticRegression.

```python
X_test_pca = model_pca.transform(X_test)
print(X_test_pca.shape)
pred = model_lr.predict(X_test_pca)


print(metrics.accuracy_score(y_test, pred))

sns.heatmap(metrics.confusion_matrix(y_test, pred), annot= True)
```

## Pipeline.


```python
pred = pipe.predict(X_test)
```

```python
print(metrics.accuracy_score(y_test, pred))

sns.heatmap(metrics.confusion_matrix(y_test, pred), annot= True)
```
